```python
import pandas as pd

# Đọc file CSV
df = pd.read_csv('tmdb-movies.csv')

# Đếm số bộ phim có đánh giá trung bình (vote_average) > 7.5
count = (df['vote_average'] > 7.5).sum()

print(f"Số bộ phim có đánh giá trung bình trên 7.5: {count}")
```

```python
import pandas as pd

# Đọc file TSV (tab-separated)
df = pd.read_csv('tmdb.tsv', sep='\t')

# Đếm số bộ phim có đánh giá trung bình (vote_average) > 7.5
count = (df['vote_average'] > 7.5).sum()

print(f"Số bộ phim có đánh giá trung bình trên 7.5: {count}")
```
